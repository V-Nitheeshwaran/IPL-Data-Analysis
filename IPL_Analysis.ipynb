```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
matches = pd.read_csv("matches.csv")
deliveries = pd.read_csv("deliveries.csv")
```

```python
matches.head()
```

```python
deliveries.head()
```

```python
matches['winner'].value_counts().plot(
    kind='bar',figsize=(10,6)
)

plt.title("Top Winning Teams in IPL")
plt.xlabel("Wins")
plt.ylabel("Teams")

plt.show()
```

```python
top_batsman = deliveries.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False).head(10)
```

```python
top_batsman.plot(kind='bar')

plt.title("Top 10 Run Scorers")
plt.xlabel("Players")
plt.ylabel("Runs")
plt.show()
```

```python
top_bowlers = deliveries['bowler'].value_counts().head(10)

top_bowlers.plot(kind='bar')

plt.title("Top 10 Bowlers")
plt.xlabel("Wickets")
plt.ylabel("Bowlers")

plt.show()
```

```python
toss_win_match_win = matches[matches['toss_winner'] == matches['winner']]

won = len(toss_win_match_win)
lost = len(matches) - won

data = [won, lost]
labels = ['Won After Toss', 'Lost After Toss']

plt.pie(
    data,
    labels=labels,
    autopct='%1.1f%%',
    wedgeprops={'width':0.4}
)

plt.title("Toss Impact on Match Result")
plt.show()
```
